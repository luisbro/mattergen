```python
from pathlib import Path
from evaluations.aggregate_metrics import aggregate
import pandas as pd

import sys

sys.path.append("./evaluations")
```

```python
df = aggregate(
    Path("generated_structures/topological_dgf2"),
    Path(
        "predictive_models/topological_mp_20/inference/predictions/topological_dgf2/predictions.csv"
    ),
)[:20_000]
```

```python
df.columns
```

```python
# different filter steps
step1 = (df["is_charge_neutral"])
step2 = (df["is_charge_neutral"] & df["unique"])
step3 = (df["is_charge_neutral"] & df["unique"] & (df["predicted_label"] == 1))
step4 = (df["is_charge_neutral"] & df["unique"] & (df["predicted_label"] == 1) & df["novel"])
```

```python
steps_df = pd.DataFrame(
    {
        "step": ["step1", "step2", "step3", "step4"],
        "criteria": ["Charge neutral", "Unique", "Predicted Topological", "Novel"],
        "num_candidates": [
            step1.sum(),
            step2.sum(),
            step3.sum(),
            step4.sum(),
        ],
    }
)
```
